```python
# ============================================================
# DECISION TREE INDUCTION
# STUDENT PERFORMANCE DATASET
# PYTHON + POWER BI DMT PROJECT
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)



# ============================================================
# 2. LOAD DATASET
# ============================================================

file_name = "student_data[1].csv"

df = pd.read_csv("/student_data[1].csv")

print("\n============================================")
print("       DATASET LOADED SUCCESSFULLY")
print("============================================")

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nFirst 7 rows:")
print(df.head())

# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\n============================================")
print("       DATASET INFORMATION")
print("============================================")

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# ============================================================
# 4. REMOVE DUPLICATES
# ============================================================

df = df.drop_duplicates()

print("\nRows after removing duplicates:", len(df))

# ============================================================
# 5. HANDLE MISSING VALUES
# ============================================================

for column in df.columns:

    if df[column].dtype == "object":

        df[column] = df[column].fillna(
            df[column].mode()[0]
        )

    else:

        df[column] = df[column].fillna(
            df[column].median()
        )

print("\nMissing values handled successfully.")

# ============================================================
# 6. CREATE PERFORMANCE CLASS
# ============================================================

# G3 is the final grade.
#
# We convert G3 into 3 classes:
#
# 0 - 9   = Low
# 10 - 14 = Medium
# 15 - 20 = High

def performance_class(mark):

    if mark <= 9:
        return "Low"

    elif mark <= 14:
        return "Medium"

    else:
        return "High"


df["Performance_Class"] = df["G3"].apply(
    performance_class
)

print("\n============================================")
print("       PERFORMANCE CLASS")
print("============================================")

print(
    df["Performance_Class"].value_counts()
)

# ============================================================
# 7. PERFORMANCE DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

df["Performance_Class"].value_counts().plot(
    kind="bar"
)

plt.title(
    "Student Performance Distribution"
)

plt.xlabel(
    "Performance Class"
)

plt.ylabel(
    "Number of Students"
)

plt.tight_layout()

plt.savefig(
    "01_Performance_Distribution.png",
    dpi=300
)

plt.show()

# ============================================================
# 8. REMOVE G3 FROM FEATURES
# ============================================================

# IMPORTANT:
#
# G3 is used to create the target.
# Therefore G3 must NOT be used as an input feature.
#
# G1 and G2 can be used as previous-period grades.

X = df.drop(
    [
        "G3",
        "Performance_Class"
    ],
    axis=1
)

y = df["Performance_Class"]

# ============================================================
# 9. ENCODE CATEGORICAL VARIABLES
# ============================================================

print("\n============================================")
print("       ENCODING CATEGORICAL DATA")
print("============================================")

label_encoders = {}

for column in X.columns:

    if X[column].dtype == "object":

        encoder = LabelEncoder()

        X[column] = encoder.fit_transform(
            X[column].astype(str)
        )

        label_encoders[column] = encoder

        print(
            column,
            "encoded successfully"
        )

# ============================================================
# 10. ENCODE TARGET
# ============================================================

target_encoder = LabelEncoder()

y_encoded = target_encoder.fit_transform(y)

print("\nTarget Classes:")

for i, class_name in enumerate(
    target_encoder.classes_
):

    print(
        i,
        "=",
        class_name
    )

# ============================================================
# 11. SAVE CLEAN DATA
# ============================================================

clean_data = X.copy()

clean_data["Performance_Class"] = y.values

clean_data["Performance_Code"] = y_encoded

clean_data.to_csv(
    "01_CLEAN_STUDENT_DATA.csv",
    index=False
)

print(
    "\nClean data saved as:",
    "01_CLEAN_STUDENT_DATA.csv"
)

# ============================================================
# 12. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\n============================================")
print("       TRAIN TEST SPLIT")
print("============================================")

print(
    "Training Records:",
    len(X_train)
)

print(
    "Testing Records:",
    len(X_test)
)

# ============================================================
# 13. CREATE DECISION TREE - GINI
# ============================================================

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42
)

# ============================================================
# 14. TRAIN DECISION TREE
# ============================================================

model.fit(
    X_train,
    y_train
)

print("\n============================================")
print("       DECISION TREE TRAINED")
print("============================================")

# ============================================================
# 15. PREDICTION
# ============================================================

y_pred = model.predict(
    X_test
)

# ============================================================
# 16. MODEL EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n============================================")
print("       MODEL PERFORMANCE")
print("============================================")

print(
    "Accuracy  :",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Precision :",
    round(precision * 100, 2),
    "%"
)

print(
    "Recall    :",
    round(recall * 100, 2),
    "%"
)

print(
    "F1 Score  :",
    round(f1 * 100, 2),
    "%"
)

# ============================================================
# 17. CLASSIFICATION REPORT
# ============================================================

print("\n============================================")
print("       CLASSIFICATION REPORT")
print("============================================")

report = classification_report(
    y_test,
    y_pred,
    target_names=target_encoder.classes_,
    zero_division=0
)

print(report)

# ============================================================
# 18. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\n============================================")
print("       CONFUSION MATRIX")
print("============================================")

print(cm)

# Save confusion matrix

cm_df = pd.DataFrame(
    cm,
    index=target_encoder.classes_,
    columns=target_encoder.classes_
)

cm_df.to_csv(
    "02_CONFUSION_MATRIX_POWERBI.csv"
)

# ============================================================
# 19. CONFUSION MATRIX GRAPH
# ============================================================

plt.figure(figsize=(7, 6))

plt.imshow(cm)

plt.title(
    "Decision Tree Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    range(len(target_encoder.classes_)),
    target_encoder.classes_
)

plt.yticks(
    range(len(target_encoder.classes_)),
    target_encoder.classes_
)

for i in range(cm.shape[0]):

    for j in range(cm.shape[1]):

        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()

plt.tight_layout()

plt.savefig(
    "02_Confusion_Matrix.png",
    dpi=300
)

plt.show()

# ============================================================
# 20. FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame({

    "Feature": X.columns,

    "Importance": model.feature_importances_

})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\n============================================")
print("       FEATURE IMPORTANCE")
print("============================================")

print(feature_importance)

feature_importance.to_csv(
    "03_FEATURE_IMPORTANCE_POWERBI.csv",
    index=False
)

# ============================================================
# 21. FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10, 8))

plt.barh(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Decision Tree Feature Importance"
)

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig(
    "03_Feature_Importance.png",
    dpi=300
)

plt.show()

# ============================================================
# 22. DECISION TREE VISUALIZATION
# ============================================================

plt.figure(
    figsize=(25, 15)
)

plot_tree(
    model,

    feature_names=X.columns,

    class_names=target_encoder.classes_,

    filled=True,

    rounded=True,

    fontsize=8
)

plt.title(
    "Decision Tree Induction - Student Performance"
)

plt.savefig(
    "04_Decision_Tree.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 23. TREE INFORMATION
# ============================================================

tree_depth = model.get_depth()

tree_leaves = model.get_n_leaves()

tree_nodes = model.tree_.node_count

print("\n============================================")
print("       DECISION TREE INFORMATION")
print("============================================")

print(
    "Tree Depth      :",
    tree_depth
)

print(
    "Number of Leaves:",
    tree_leaves
)

print(
    "Number of Nodes :",
    tree_nodes
)

# ============================================================
# 24. ACTUAL VS PREDICTED
# ============================================================

actual_classes = target_encoder.inverse_transform(
    y_test
)

predicted_classes = target_encoder.inverse_transform(
    y_pred
)

testing_results = pd.DataFrame({

    "Actual_Class":
        actual_classes,

    "Predicted_Class":
        predicted_classes

})

testing_results["Correct"] = (
    testing_results["Actual_Class"]
    ==
    testing_results["Predicted_Class"]
)

testing_results.to_csv(
    "04_ACTUAL_VS_PREDICTED_POWERBI.csv",
    index=False
)

print(
    "\nActual vs Predicted data saved."
)

# ============================================================
# 25. MODEL METRICS FOR POWER BI
# ============================================================

metrics_df = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score",

        "Tree Depth",

        "Tree Leaves",

        "Tree Nodes"

    ],

    "Value": [

        accuracy,

        precision,

        recall,

        f1,

        tree_depth,

        tree_leaves,

        tree_nodes

    ]

})

metrics_df.to_csv(
    "05_MODEL_METRICS_POWERBI.csv",
    index=False
)

# ============================================================
# 26. GINI VS ENTROPY
# ============================================================

print("\n============================================")
print("       GINI VS ENTROPY")
print("============================================")

# GINI

gini_model = DecisionTreeClassifier(

    criterion="gini",

    max_depth=5,

    min_samples_split=10,

    min_samples_leaf=5,

    random_state=42

)

gini_model.fit(
    X_train,
    y_train
)

gini_prediction = gini_model.predict(
    X_test
)

gini_accuracy = accuracy_score(
    y_test,
    gini_prediction
)

# ENTROPY

entropy_model = DecisionTreeClassifier(

    criterion="entropy",

    max_depth=5,

    min_samples_split=10,

    min_samples_leaf=5,

    random_state=42

)

entropy_model.fit(
    X_train,
    y_train
)

entropy_prediction = entropy_model.predict(
    X_test
)

entropy_accuracy = accuracy_score(
    y_test,
    entropy_prediction
)

comparison = pd.DataFrame({

    "Criterion": [

        "Gini",

        "Entropy"

    ],

    "Accuracy": [

        gini_accuracy,

        entropy_accuracy

    ],

    "Tree_Depth": [

        gini_model.get_depth(),

        entropy_model.get_depth()

    ],

    "Leaves": [

        gini_model.get_n_leaves(),

        entropy_model.get_n_leaves()

    ]

})

print(comparison)

comparison.to_csv(
    "06_GINI_VS_ENTROPY_POWERBI.csv",
    index=False
)

# ============================================================
# 27. TREE DEPTH COMPARISON
# ============================================================

print("\n============================================")
print("       TREE DEPTH COMPARISON")
print("============================================")

depth_results = []

for depth in [2, 3, 4, 5, 6, 7, 8]:

    temp_model = DecisionTreeClassifier(

        criterion="gini",

        max_depth=depth,

        random_state=42

    )

    temp_model.fit(
        X_train,
        y_train
    )

    temp_prediction = temp_model.predict(
        X_test
    )

    temp_accuracy = accuracy_score(
        y_test,
        temp_prediction
    )

    depth_results.append({

        "Max_Depth": depth,

        "Accuracy": temp_accuracy,

        "Actual_Depth":
            temp_model.get_depth(),

        "Leaves":
            temp_model.get_n_leaves()

    })

depth_df = pd.DataFrame(
    depth_results
)

print(depth_df)

depth_df.to_csv(
    "07_TREE_DEPTH_COMPARISON_POWERBI.csv",
    index=False
)

# ============================================================
# 28. FULL DATASET PREDICTION
# ============================================================

all_predictions = model.predict(X)

all_prediction_classes = (
    target_encoder.inverse_transform(
        all_predictions
    )
)

powerbi_data = df.copy()

powerbi_data[
    "Predicted_Class"
] = all_prediction_classes

powerbi_data[
    "Actual_Class_Code"
] = y_encoded

powerbi_data[
    "Predicted_Class_Code"
] = all_predictions

powerbi_data[
    "Prediction_Correct"
] = (

    powerbi_data[
        "Performance_Class"
    ]

    ==

    powerbi_data[
        "Predicted_Class"
    ]

)

powerbi_data.to_csv(
    "08_FINAL_POWERBI_DATA.csv",
    index=False
)

# ============================================================
# 29. G3 DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 6))

plt.hist(
    df["G3"],
    bins=range(
        int(df["G3"].min()),
        int(df["G3"].max()) + 2
    )
)

plt.title(
    "Final Grade (G3) Distribution"
)

plt.xlabel(
    "Final Grade"
)

plt.ylabel(
    "Number of Students"
)

plt.tight_layout()

plt.savefig(
    "05_G3_Distribution.png",
    dpi=300
)

plt.show()

# ============================================================
# 30. G1 VS G2 VS G3
# ============================================================

average_marks = [

    df["G1"].mean(),

    df["G2"].mean(),

    df["G3"].mean()

]

plt.figure(figsize=(7, 5))

plt.bar(
    ["G1", "G2", "G3"],
    average_marks
)

plt.title(
    "Average Student Marks"
)

plt.xlabel(
    "Assessment"
)

plt.ylabel(
    "Average Marks"
)

plt.tight_layout()

plt.savefig(
    "06_Average_Marks.png",
    dpi=300
)

plt.show()

# ============================================================
# 31. STUDY TIME VS PERFORMANCE
# ============================================================

study_performance = df.groupby(
    "studytime"
)["G3"].mean()

plt.figure(figsize=(8, 5))

plt.plot(
    study_performance.index,
    study_performance.values,
    marker="o"
)

plt.title(
    "Study Time vs Average Final Grade"
)

plt.xlabel(
    "Study Time"
)

plt.ylabel(
    "Average G3"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "07_Study_Time_vs_Performance.png",
    dpi=300
)

plt.show()

# ============================================================
# 32. ABSENCES VS PERFORMANCE
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["absences"],
    df["G3"]
)

plt.title(
    "Absences vs Final Grade"
)

plt.xlabel(
    "Absences"
)

plt.ylabel(
    "Final Grade (G3)"
)

plt.tight_layout()

plt.savefig(
    "08_Absences_vs_Performance.png",
    dpi=300
)

plt.show()

# ============================================================
# 33. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("====================================================")
print("       DECISION TREE PROJECT COMPLETED")
print("====================================================")

print(
    "\nDataset Records:",
    len(df)
)

print(
    "Dataset Columns:",
    len(df.columns)
)

print(
    "\nAccuracy:",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Precision:",
    round(precision * 100, 2),
    "%"
)

print(
    "Recall:",
    round(recall * 100, 2),
    "%"
)

print(
    "F1 Score:",
    round(f1 * 100, 2),
    "%"
)

print(
    "\nTree Depth:",
    tree_depth
)

print(
    "Tree Leaves:",
    tree_leaves
)

print(
    "Tree Nodes:",
    tree_nodes
)

print("\n====================================================")
print("             FILES CREATED")
print("====================================================")

files = [

    "01_CLEAN_STUDENT_DATA.csv",

    "02_CONFUSION_MATRIX_POWERBI.csv",

    "03_FEATURE_IMPORTANCE_POWERBI.csv",

    "04_ACTUAL_VS_PREDICTED_POWERBI.csv",

    "05_MODEL_METRICS_POWERBI.csv",

    "06_GINI_VS_ENTROPY_POWERBI.csv",

    "07_TREE_DEPTH_COMPARISON_POWERBI.csv",

    "08_FINAL_POWERBI_DATA.csv",

    "01_Performance_Distribution.png",

    "02_Confusion_Matrix.png",

    "03_Feature_Importance.png",

    "04_Decision_Tree.png",

    "05_G3_Distribution.png",

    "06_Average_Marks.png",

    "07_Study_Time_vs_Performance.png",

    "08_Absences_vs_Performance.png"

]

for i, file in enumerate(files, 1):

    print(
        i,
        ".",
        file
    )

print("\n====================================================")
print("             READY FOR POWER BI")
print("====================================================")
```
